# Healthcare Data Analysis Project 

**Dataset:** 10,000+ patient health records  
**Tools:** Python, Pandas, NumPy, Matplotlib, Seaborn

## Resume-aligned objectives
- Conduct EDA on 10,000+ patient health records to identify key health trends.
- Scrutinize healthcare datasets to support insights related to patient-care improvement.
- Identify patterns in disease occurrence and treatment outcomes.
  

> **Note:** This is a synthetic industry-style healthcare dataset created for training/portfolio practice. It does not contain real patient information.


```python
# Lab 1 – Import libraries and load the dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

file_path = "healthcare_patient_records_10000.csv"
df = pd.read_csv(file_path)
df
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
df.head()
```

```python
# Lab 2 – Data structure and data quality
df.info()
```

```python
# Missing values and duplicate records
print("Missing values:")
display(df.isnull().sum().sort_values(ascending=False))
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# Lab 3 – Data cleaning
df = df.drop_duplicates().copy()

numeric_cols = [
    "BMI", "Glucose_mg_dL", "Total_Cholesterol_mg_dL",
    "Patient_Satisfaction_Score"
]

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

df["Admission_Date"] = pd.to_datetime(df["Admission_Date"])

print("Rows after cleaning:", len(df))
print("Remaining missing values:", df.isnull().sum().sum())
```

```python
# Lab 4 – Descriptive statistics
display(df.describe(include="all").T)
```

```python
# Lab 5 – Patient demographic analysis
print("Average age:", round(df["Age"].mean(), 2))
print("Male patients:", (df["Gender"]=="Male").sum())
print("Female patients:", (df["Gender"]=="Female").sum())

display(df["Gender"].value_counts().rename_axis("Gender").reset_index(name="Patients"))
```

```python
# Lab 6 – Disease occurrence analysis
disease_summary = (
    df["Primary_Diagnosis"]
    .value_counts()
    .rename_axis("Primary_Diagnosis")
    .reset_index(name="Patient_Count")
)
disease_summary["Percentage"] = (
    disease_summary["Patient_Count"] / len(df) * 100
).round(2)

display(disease_summary)
```

```python
# Visualize disease occurrence
plt.figure(figsize=(11,5))
sns.countplot(data=df, y="Primary_Diagnosis",
              order=df["Primary_Diagnosis"].value_counts().index)
plt.title("Disease Occurrence by Diagnosis")
plt.xlabel("Number of Patients")
plt.ylabel("Diagnosis")
plt.show()
```

```python
# Lab 7 – Comorbidity analysis
comorbidity = df.groupby(["Hypertension", "Diabetes", "Heart_Disease"]).size().reset_index(name="Patients")
display(comorbidity.sort_values("Patients", ascending=False).head(10))
```

```python
# Lab 8 – Clinical measurements
clinical_cols = [
    "Age", "BMI", "Systolic_BP", "Diastolic_BP",
    "Heart_Rate", "Glucose_mg_dL", "Total_Cholesterol_mg_dL"
]
display(df[clinical_cols].describe().T)
```

```python
# Lab 9 – Correlation analysis
corr = df[clinical_cols].corr()

plt.figure(figsize=(9,7))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Clinical Variable Correlation")
plt.show()
```

```python
# Lab 10 – Treatment outcome analysis
outcome_summary = (
    df["Treatment_Outcome"]
    .value_counts()
    .rename_axis("Treatment_Outcome")
    .reset_index(name="Patients")
)
outcome_summary["Percentage"] = (outcome_summary["Patients"]/len(df)*100).round(2)
display(outcome_summary)
```

```python
# Treatment outcome by disease
outcome_by_disease = pd.crosstab(
    df["Primary_Diagnosis"], df["Treatment_Outcome"], normalize="index"
).mul(100).round(2)

display(outcome_by_disease)
```

```python
# Lab 11 – Length of stay analysis
los_summary = (
    df.groupby("Primary_Diagnosis")["Length_of_Stay_Days"]
      .agg(["count","mean","median","max"])
      .sort_values("mean", ascending=False)
)

display(los_summary)
```

```python
# Lab 12 – Readmission analysis
readmission_summary = (
    df["Readmission_30_Days"]
    .value_counts(normalize=True)
    .mul(100).round(2)
)
display(readmission_summary)
```

```python
# Readmission rate by diagnosis
readmission_by_disease = (
    df.groupby("Primary_Diagnosis")["Readmission_30_Days"]
      .apply(lambda x: (x=="Yes").mean()*100)
      .sort_values(ascending=False)
      .round(2)
      .reset_index(name="Readmission_Rate_%")
)
display(readmission_by_disease)
```

```python
# Lab 13 – Treatment cost analysis
cost_summary = (
    df.groupby(["Treatment_Type", "Treatment_Outcome"])["Treatment_Cost_USD"]
      .agg(["count","mean","median"])
      .sort_values("mean", ascending=False)
)

display(cost_summary)
```

```python
# Lab 14 – Patient satisfaction and care improvement
care_summary = (
    df.groupby("Department")
      .agg(
          Patients=("Patient_ID","count"),
          Avg_Satisfaction=("Patient_Satisfaction_Score","mean"),
          Avg_LOS=("Length_of_Stay_Days","mean"),
          Readmission_Rate=("Readmission_30_Days", lambda x: (x=="Yes").mean()*100)
      )
      .sort_values("Readmission_Rate", ascending=False)
)

care_summary["Avg_Satisfaction"] = care_summary["Avg_Satisfaction"].round(2)
care_summary["Avg_LOS"] = care_summary["Avg_LOS"].round(2)
care_summary["Readmission_Rate"] = care_summary["Readmission_Rate"].round(2)

display(care_summary)
```

```python
# Lab 15 – Monthly patient trend
df["Admission_Month"] = df["Admission_Date"].dt.to_period("M").astype(str)

monthly = (
    df.groupby("Admission_Month")
      .agg(
          Patients=("Patient_ID","count"),
          Avg_LOS=("Length_of_Stay_Days","mean"),
          Avg_Cost=("Treatment_Cost_USD","mean")
      )
      .reset_index()
)

display(monthly.head())

plt.figure(figsize=(12,5))
sns.lineplot(data=monthly, x="Admission_Month", y="Patients", marker="o")
plt.xticks(rotation=60)
plt.title("Monthly Patient Admission Trend")
plt.xlabel("Month")
plt.ylabel("Patients")
plt.tight_layout()
plt.show()
```

```python
# Lab 16 – Executive healthcare KPIs
total_patients = len(df)
avg_age = df["Age"].mean()
avg_los = df["Length_of_Stay_Days"].mean()
recovery_rate = (df["Treatment_Outcome"]=="Recovered").mean()*100
readmission_rate = (df["Readmission_30_Days"]=="Yes").mean()*100
avg_satisfaction = df["Patient_Satisfaction_Score"].mean()
avg_cost = df["Treatment_Cost_USD"].mean()

kpis = pd.DataFrame({
    "KPI": [
        "Total Patients", "Average Age", "Average Length of Stay",
        "Recovery Rate (%)", "30-Day Readmission Rate (%)",
        "Average Satisfaction", "Average Treatment Cost"
    ],
    "Value": [
        total_patients, round(avg_age,2), round(avg_los,2),
        round(recovery_rate,2), round(readmission_rate,2),
        round(avg_satisfaction,2), round(avg_cost,2)
    ]
})
display(kpis)
```
